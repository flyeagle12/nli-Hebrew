```python
import pandas as pd

test_path = "./allData/HebNLI_test.full.clean.jsonl"
df = pd.read_json(test_path, lines=True)

# ========================================
# SIMPLE LABEL SPLIT VIEW
# ========================================

print("\n" + "="*70)
print("LABEL DISTRIBUTION")
print("="*70)

# Count
label_counts = df["original_label"].value_counts().sort_index()
print("\nCounts:")
print(label_counts)

# Percentage
label_pcts = df["original_label"].value_counts(normalize=True).sort_index() * 100
print("\nPercentages:")
for label, pct in label_pcts.items():
    count = label_counts[label]
    print(f"  {label:15s}: {count:5d} ({pct:5.1f}%)")

print(f"\nTotal: {len(df)}")

# ========================================
# FOR YOUR BALANCED SET
# ========================================

pivot = pd.crosstab(df["original_label"], df["genre"])
min_count = int(pivot.min().min())

parts = []
for (lab, gen), gdf in df.groupby(["original_label", "genre"]):
    parts.append(gdf.sample(n=min_count, random_state=42))

df_bal = pd.concat(parts).sample(frac=1, random_state=42).reset_index(drop=True)

print("\n" + "="*70)
print("BALANCED SET LABEL DISTRIBUTION")
print("="*70)

label_counts_bal = df_bal["original_label"].value_counts().sort_index()
print("\nCounts:")
print(label_counts_bal)

label_pcts_bal = df_bal["original_label"].value_counts(normalize=True).sort_index() * 100
print("\nPercentages:")
for label, pct in label_pcts_bal.items():
    count = label_counts_bal[label]
    print(f"  {label:15s}: {count:5d} ({pct:5.1f}%)")

print(f"\nTotal: {len(df_bal)}")

# ========================================
# FOR YOUR TRAIN HALF
# ========================================

half_size = len(df) // 2
df_half = df.iloc[:half_size]

print("\n" + "="*70)
print("TRAIN HALF LABEL DISTRIBUTION")
print("="*70)

label_counts_train = df_half["original_label"].value_counts().sort_index()
print("\nCounts:")
print(label_counts_train)

label_pcts_train = df_half["original_label"].value_counts(normalize=True).sort_index() * 100
print("\nPercentages:")
for label, pct in label_pcts_train.items():
    count = label_counts_train[label]
    print(f"  {label:15s}: {count:5d} ({pct:5.1f}%)")

print(f"\nTotal: {len(df_half)}")

# ========================================
# SIDE-BY-SIDE COMPARISON
# ========================================

print("\n" + "="*70)
print("SIDE-BY-SIDE COMPARISON")
print("="*70)

comparison = pd.DataFrame({
    'Original_Count': label_counts,
    'Original_%': label_pcts.round(1),
    'Balanced_Count': label_counts_bal,
    'Balanced_%': label_pcts_bal.round(1),
    'Train_Count': label_counts_train,
    'Train_%': label_pcts_train.round(1)
})

print("\n", comparison)
```

```python
import pandas as pd
import json

test_path = "./allData/HebNLI_test.full.clean.jsonl"

# Load data
df = pd.read_json(test_path, lines=True)

print("="*70)
print("ORIGINAL DATA ANALYSIS")
print("="*70)

# 1. Overall label distribution
print("\n📊 LABEL DISTRIBUTION (Overall):")
print(df["original_label"].value_counts().sort_index())
print("\nPercentages:")
print(df["original_label"].value_counts(normalize=True).sort_index() * 100)

# 2. Label × Genre crosstab
print("\n" + "="*70)
print("LABEL × GENRE CROSSTAB")
print("="*70)
pivot = pd.crosstab(df["original_label"], df["genre"], margins=True)
print(pivot)

# 3. Percentages within each genre
print("\n" + "="*70)
print("LABEL DISTRIBUTION WITHIN EACH GENRE (%)")
print("="*70)
pivot_pct = pd.crosstab(df["original_label"], df["genre"], normalize='columns') * 100
print(pivot_pct.round(1))

# 4. Percentages within each label
print("\n" + "="*70)
print("GENRE DISTRIBUTION WITHIN EACH LABEL (%)")
print("="*70)
pivot_pct2 = pd.crosstab(df["original_label"], df["genre"], normalize='index') * 100
print(pivot_pct2.round(1))

# ========================================
# BALANCED TEST SET
# ========================================
print("\n\n" + "="*70)
print("CREATING BALANCED TEST SET")
print("="*70)

min_count = int(pivot.min().min())
print(f"\n✓ Min count per label×genre cell: {min_count}")

parts = []
for (lab, gen), gdf in df.groupby(["original_label", "genre"]):
    sampled = gdf.sample(n=min_count, random_state=42)
    parts.append(sampled)
    print(f"  Sampled {min_count} from {lab:15s} × {gen}")

df_bal = pd.concat(parts).sample(frac=1, random_state=42).reset_index(drop=True)

print(f"\n✓ Balanced test rows: {len(df_bal)}")

print("\n📊 BALANCED LABEL DISTRIBUTION:")
print(df_bal["original_label"].value_counts().sort_index())
print("\nPercentages:")
print(df_bal["original_label"].value_counts(normalize=True).sort_index() * 100)

print("\n" + "="*70)
print("BALANCED LABEL × GENRE CROSSTAB")
print("="*70)
pivot_bal = pd.crosstab(df_bal["original_label"], df_bal["genre"], margins=True)
print(pivot_bal)

# ========================================
# TRAIN SET (HALF OF DATA)
# ========================================
print("\n\n" + "="*70)
print("TRAIN SET ANALYSIS (First Half)")
print("="*70)

half_size = len(df) // 2
df_half = df.iloc[:half_size]

print(f"\n✓ Train set size: {len(df_half)} rows")

print("\n📊 TRAIN LABEL DISTRIBUTION:")
print(df_half["original_label"].value_counts().sort_index())
print("\nPercentages:")
print(df_half["original_label"].value_counts(normalize=True).sort_index() * 100)

pivot_train = pd.crosstab(df_half["original_label"], df_half["genre"], margins=True)
min_count_train = int(pivot_train.iloc[:-1, :-1].min().min())  # Exclude margins

print(f"\n✓ Train min label×genre cell: {min_count_train}")

print("\n" + "="*70)
print("TRAIN LABEL × GENRE CROSSTAB")
print("="*70)
print(pivot_train)

# ========================================
# COMPARISON SUMMARY
# ========================================
print("\n\n" + "="*70)
print("SUMMARY COMPARISON")
print("="*70)

summary = pd.DataFrame({
    'Original': df["original_label"].value_counts().sort_index(),
    'Balanced_Test': df_bal["original_label"].value_counts().sort_index(),
    'Train_Half': df_half["original_label"].value_counts().sort_index()
})

print("\nLabel counts across splits:")
print(summary)

print("\nLabel percentages across splits:")
summary_pct = summary / summary.sum() * 100
print(summary_pct.round(1))

# ========================================
# SAVE RESULTS
# ========================================
print("\n" + "="*70)
print("SAVING RESULTS")
print("="*70)

# Save balanced test set
output_path = "./allData/HebNLI_test.balanced.jsonl"
df_bal.to_json(output_path, orient='records', lines=True, force_ascii=False)
print(f"✓ Balanced test saved to: {output_path}")

# Save summary report
report_path = "./allData/label_split_report.txt"
with open(report_path, 'w', encoding='utf-8') as f:
    f.write("HEBREW NLI DATASET - LABEL SPLIT ANALYSIS\n")
    f.write("="*70 + "\n\n")
    
    f.write("ORIGINAL DATA:\n")
    f.write(f"Total: {len(df)} examples\n")
    f.write(df["original_label"].value_counts().sort_index().to_string())
    f.write("\n\n")
    
    f.write("BALANCED TEST SET:\n")
    f.write(f"Total: {len(df_bal)} examples\n")
    f.write(df_bal["original_label"].value_counts().sort_index().to_string())
    f.write("\n\n")
    
    f.write("TRAIN SET (First Half):\n")
    f.write(f"Total: {len(df_half)} examples\n")
    f.write(df_half["original_label"].value_counts().sort_index().to_string())
    f.write("\n\n")
    
    f.write("LABEL × GENRE CROSSTAB (Original):\n")
    f.write(pivot.to_string())

print(f"✓ Report saved to: {report_path}")

print("\n" + "="*70)
print("ANALYSIS COMPLETE")
print("="*70)
```
