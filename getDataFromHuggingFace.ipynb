```python
import requests
import pandas as pd
import time
```

```python
def download_full_split(dataset, split="train"):
    base_url = "https://datasets-server.huggingface.co/rows"
    all_rows = []
    offset = 0
    length = 100
    
    print(f"Starting download for '{dataset}' (Split: {split})...")
    
    while True:
        params = {
            "dataset": dataset,
            "config": "default",
            "split": split,
            "offset": offset,
            "length": length
        }
        
        try:
            response = requests.get(base_url, params=params)
            
            # Check if the request was successful
            if response.status_code != 200:
                print(f"Error: API returned status {response.status_code}")
                print(response.text)
                break
                
            data = response.json()
            
            # Extract the rows
            batch = [item['row'] for item in data['rows']]
            
            # If the batch is empty, we have reached the end
            if not batch:
                break
                
            all_rows.extend(batch)
            
            # Print progress
            print(f"Downloaded {len(all_rows)} rows so far...", end="\r")
            
            # increment offset for next page
            offset += length
            
            # Be polite to the server
            time.sleep(0.1)
            
        except Exception as e:
            print(f"Failed at offset {offset}: {e}")
            break
            
    print(f"\nFinished! Total rows: {len(all_rows)}")
    return pd.DataFrame(all_rows)

# --- EXECUTE ---
df_train = download_full_split("HebArabNlpProject/HebNLI", split="train")

# Show result
print(df_train.head())
```
