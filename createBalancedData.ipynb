```python
import pandas as pd
import json

train_path = "./allData/HebNLI_train.full.clean.jsonl"
out_path = "./allData/HebNLI_train.balanced.jsonl"

df = pd.read_json(train_path, lines=True)

pivot = pd.crosstab(df["original_label"], df["genre"])
min_count = int(pivot.min().min())
print("Train min label×genre cell =", min_count)  # should print 563

parts = []
for (lab, gen), gdf in df.groupby(["original_label", "genre"]):
    parts.append(gdf.sample(n=min_count, random_state=42))

df_bal = pd.concat(parts).sample(frac=1, random_state=42).reset_index(drop=True)
print("Balanced train rows:", len(df_bal))  # 3*7*563 = 11823

with open(out_path, "w", encoding="utf-8") as f:
    for rec in df_bal.to_dict(orient="records"):
        f.write(json.dumps(rec, ensure_ascii=False) + "\n")

print("Saved:", out_path)
```

```python
import pandas as pd
import json

train_path = "./allData/HebNLI_train.full.clean.jsonl"

df = pd.read_json(train_path, lines=True)
half_size = len(df) // 2
df_half = df.iloc[:half_size]

half_path = "./allData/HebNLI_train.half.jsonl"
with open(half_path, "w", encoding="utf-8") as f:
    for rec in df_half.to_dict(orient="records"):
        f.write(json.dumps(rec, ensure_ascii=False) + "\n")

print(f"Saved half of the data ({len(df_half)} rows) to:", half_path)
```
